 Business Scenario
Sales team enters daily sales values. Management wants total & average sales.

```python
days = int(input("Enter days: "))
total = 0

for i in range(days):
    total += float(input("Enter sales: "))

average = total / days

print("Total Sales:", total)
print("Average Sales:", average)
```
