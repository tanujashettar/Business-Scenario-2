 Attendance Summary Generator
(for loop)


Business Scenario
Trainer records attendance for interns.

```python
n = int(input("Enter number of interns: "))
present = 0
absent = 0

for i in range(n):
    status = input("Enter attendance (P/A): ").upper()

    if status == "P":
        present += 1
    else:
        absent += 1

print("Total Present:", present)
print("Total Absent:", absent)
```
