Inventory Reorder Checker
(for loop + conditional logic)

Business Scenario
Warehouse checks which products need reorder.

```python
n = int(input("Enter number of products: "))

for i in range(1, n + 1):
    stock = int(input(f"Enter stock for product {i}: "))

    if stock < 20:
        print(f"Product {i}: Reorder Required")
    else:
        print(f"Product {i}: Stock Sufficient")
```
