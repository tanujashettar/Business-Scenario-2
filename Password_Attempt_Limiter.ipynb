Password Attempt Limiter
(while loop)

Business Scenario
System limits number of password attempts.

```python
password = "admin123"
attempts = 0

while attempts < 3:
    user_pass = input("Enter password: ")

    if user_pass == password:
        print("Login Successful")
        break
    else:
        attempts += 1

if attempts == 3:
    print("Account Locked")
```
